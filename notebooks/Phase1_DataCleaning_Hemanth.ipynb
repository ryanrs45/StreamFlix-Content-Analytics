## Phase 1 – Data Quality Summary

### 1. subscribers.csv
- Total records: 15,000
- Total columns: 14
- Duplicate subscriber IDs: 0
- Fully duplicated rows: 0
- Missing values were found only in `churn_date`.
- `churn_date` has 11,199 missing values, representing 74.66% of the table.
- The missing churn dates are expected for active subscribers.
- No active subscribers had a churn date.
- No inactive subscribers were missing a churn date.
- All churn dates occurred after the corresponding signup dates.
- `signup_date` and `churn_date` were converted to datetime format.
- Subscriber ages ranged from 13 to 85 years.
- The average subscriber age was approximately 35.62 years.
- All plan types were valid: Basic with Ads, Standard, and Premium.

### 2. titles.csv
- Total records: 9,000
- Total columns: 21
- Duplicate title IDs: 0
- Fully duplicated rows: 0
- Missing values were found only in `license_expiry`.
- `license_expiry` has 1,966 missing values, representing approximately 21.84% of the table.
- All missing licence-expiry values belonged to StreamFlix Original titles, so these null values are valid.
- `date_added` and `license_expiry` were converted to datetime format.
- The dataset contains 5,336 Movies and 3,664 TV Shows.
- All Movies correctly have `seasons = 0`.
- All TV Shows have at least one season.

### 3. watch_history.csv
- Total records: 650,000
- Total columns: 10
- Missing values: 0
- Duplicate watch IDs: 0

- The maximum completion-percentage difference was approximately 0.16 percentage points.

- Every subscriber ID exists in the subscribers table.

### Overall Data Quality Conclusion
The StreamFlix datasets are generally high quality. No duplicate primary keys, full duplicate rows, invalid foreign-key relationships, or major logical inconsistencies were identified. Missing values in `churn_date` and `license_expiry` are expected based on the business rules of the dataset. Date columns were converted to proper datetime formats, and the cleaned datasets are ready for exploratory data analysis and KPI calculations.
